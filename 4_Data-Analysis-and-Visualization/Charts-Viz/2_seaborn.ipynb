```python
import seaborn as sns
import matplotlib.pyplot as plt
penguins = sns.load_dataset("penguins").dropna()
```

```python
# 1. Scatter plot
sns.scatterplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', hue='species')
plt.show()
```

```python
# 2. Line plot
sns.lineplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', hue='species')
plt.show()
```

```python
# 3. Histogram
sns.histplot(data=penguins, x='body_mass_g', hue='species', kde=True)
plt.show()
```

```python
# 4. KDE plot (smooth distribution curve)
sns.kdeplot(data=penguins, x='body_mass_g', hue='species', fill=True)
plt.show()
```

```python
# 5. Bar plot (mean + confidence interval per category)
sns.barplot(data=penguins, x='species', y='body_mass_g')
plt.show()
```

```python
# 6. Count plot (frequency of categories)
sns.countplot(data=penguins, x='species', hue='island')
plt.show()
```

```python
# 7. Box plot
sns.boxplot(data=penguins, x='species', y='flipper_length_mm')
plt.show()
```

```python
# 8. Violin plot
sns.violinplot(data=penguins, x='species', y='flipper_length_mm')
plt.show()
```

```python
# 9. Strip plot (individual points, jittered)
sns.stripplot(data=penguins, x='species', y='bill_length_mm', jitter=True)
plt.show()
```

```python
# 10. Swarm plot (individual points, non-overlapping)
sns.swarmplot(data=penguins, x='species', y='bill_length_mm')
plt.show()
```

```python
# 11. Pair plot (all numeric vars vs each other)
sns.pairplot(penguins, hue='species')
plt.show()
```

```python
# 12. Heatmap (correlation matrix)
sns.heatmap(penguins.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

```python
# 13. Joint plot (scatter + marginal histograms)
sns.jointplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', hue='species')
plt.show()
```

```python
# 14. Reg plot (scatter + regression line)
sns.regplot(data=penguins, x='bill_length_mm', y='bill_depth_mm')
plt.show()
```

```python
# 15. FacetGrid (split into subplots by category)
g = sns.FacetGrid(penguins, col='species')
g.map(sns.histplot, 'body_mass_g')
plt.show()
```

```python
# 16. Cat plot (generic categorical plotter — kind='box','violin','bar','strip','swarm', etc.)
sns.catplot(data=penguins, x='species', y='body_mass_g', kind='box')
plt.show()
```
