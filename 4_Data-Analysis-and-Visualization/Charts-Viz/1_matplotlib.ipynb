```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns #only for dataset
df = sns.load_dataset("penguins")
```

## Basic Dataset

```python
#Line Plot
x = [1,2,3,4,5]
y = [1,4,9,1,25]
plt.plot(x,y,color='red', linestyle='--', marker='o', linewidth=2, markersize=9)
plt.xlabel('Numbers')
plt.ylabel('Squares')
plt.title('Basic Line Plot of Numbers and Squares')
plt.grid(True)
plt.show()
```

```python
df.info()
```

```python
df['species'].value_counts()
```

```python
#Scatter Plot - Compare Relationships b/w 2 variables
specie_colors = {'Adelie': 'orange', 'Chinstrap': 'green', 'Gentoo': 'blue'}
for species, colors in specie_colors.items():
    subset = df[df['species']==species]
    plt.scatter(subset['bill_length_mm'], subset['bill_depth_mm'], label=species, color=colors)
plt.xlabel('Bill Length (mm)')
plt.ylabel('Bill Depth (mm)')
plt.title('Bill Length vs Depth by Species')
plt.legend()
plt.show()
```

```python
# Histogram
plt.hist(df['body_mass_g'], bins=20, edgecolor='black')
plt.xlabel('Body Mass (g)')
plt.ylabel('Frequency')
plt.title('Distribution of Body Mass')
plt.show()
```

```python
#Box Plot
species_list = df['species'].unique()
data = [df[df['species'] == s]['flipper_length_mm'] for s in species_list]

plt.boxplot(data, labels=species_list)
plt.ylabel('Flipper Length (mm)')
plt.title('Flipper Length by Species')
plt.show()
```

```python
#Bar Chart - we usually group categories here
avg_mass = df.groupby('species')['body_mass_g'].mean()
plt.bar(avg_mass.index, avg_mass.values, color = ['orange', 'green', 'blue'])
plt.xlabel('Species')
plt.ylabel('Average Body Mass (g)')
plt.title('Average Body Mass by Species')
plt.show()
```

## ACTUAL DATASET
Dataset of Immigrations in Canada by each Country

```python
df_can = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Canada.csv')
df_can.head()
```

```python
df_can.set_index('Country', inplace=True)
df_can.head()
```

```python
df_can.info()
```

```python
years = list()
for year in range(1980,2014):
    years.append(year)
print(years, end='')
```

Draw Line Plot of Haiti Immigrants each year

```python
haiti = df_can.loc['Haiti']
haiti_years = haiti.drop(['Continent', 'Region', 'DevName', 'Total'])
haiti_years
```

```python
haiti_years.plot(kind='line', marker='o')
plt.xlabel('Years')
plt.ylabel('Immigrants')
plt.title('Haiti Immigrants Each year')
plt.show()
```

```python
haiti_years.plot(kind='line', marker='o')
plt.text(2000,6000, '2010 EarthQuake -->')
plt.show()
```

```python
#Repeat for India+China
df_CI = df_can.loc[['India', 'China']]
df_CI=df_CI.transpose()
CI_years = df_CI.drop(['Continent', 'Region', 'DevName', 'Total'])
```

```python
CI_years.plot(kind='line', marker='o')
plt.xlabel('Years')
plt.ylabel('Immigrants')
plt.title('India+China Immigrants Each year')
plt.show()
```

```python
## Top 5 Countries that immigrated in Canada
df_can.sort_values(by='Total', ascending=False, inplace=True)
top5 = df_can.head()
top5 = top5.drop(columns=['Continent', 'Region', 'DevName', 'Total']).transpose()
top5.rename(columns={'United Kingdom of Great Britain and Northern Ireland': 'UK'}, inplace=True)
top5
```

```python
top5.plot(kind='line', figsize=(10,6))
plt.xlabel('Years')
plt.ylabel('Immigrants')
plt.title('Top 5 Countries by Immigrants Each year')
plt.legend(loc='center left',bbox_to_anchor=(1.0, 0.5))
plt.tight_layout()
plt.show()
```

## DIFFICULT PLOTS

```python
## Area Plot
top5.plot(kind='area', stacked=False, figsize=(20,10),alpha=0.25)
plt.title('Immigration Trend of Top 5 Countries')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

plt.show()
```

```python
##Above Style is called scripting, we have another called artistic which is object oriented
ax = top5.plot(kind='area', stacked=False, figsize=(20,10),alpha=0.25)
ax.set_title('Immigration Trend of Top 5 Countries')
ax.set_ylabel('Number of Immigrants')
ax.set_xlabel('Years')
```

```python
## Bar Graph
df_can_2013 = df_can['2013']
df_can_2013.sort_values(inplace=True, ascending=False)
df_can_2013=df_can_2013.head()
df_can_2013
```

```python
df_can_2013.plot(kind='bar',rot=45)
plt.title('Immigration from Top 5 Countries in 2013')
plt.xlabel('Country')
plt.ylabel('Number of Immigrants')
pak_pos = df_can_2013.index.get_loc('Pakistan')
plt.show()
```

```python
## Histogram
count, bin_edges = np.histogram(df_can['2013'])

df_can['2013'].plot(kind='hist', bins=bin_edges ,figsize=(10,6), edgecolor='black')
plt.title('Distribution of Immigration Across All Countries in 2013')
plt.xticks(bin_edges)
plt.xlabel('Number of Immigrants')
plt.ylabel('Number of Countries')
plt.show()
```

```python
## Subplots - multiple plots in 1 fig
fig, axs = plt.subplots(2,2,figsize=(14,10))

df_can_2013.plot(kind='hist', ax=axs[0,0], edgecolor='black')
axs[0,0].set_title('Histogram')

top5.plot(kind='line', ax=axs[0,1])
axs[0,1].set_title('Line Plot')

df_can_2013.plot(kind='bar', ax=axs[1,0], rot=45)
axs[1,0].set_title('Bar Chart')

df.boxplot(column='flipper_length_mm', by='species', ax=axs[1,1])
axs[1,1].set_title('Box Plot')

plt.tight_layout()
plt.show()
```
